```python
import glob
import pandas as pd

files = glob.glob("/kaggle/input/**/*.csv", recursive=True)
print(files)
```

```python
df=pd.read_csv(files[0],dtype=str)
print("Number of puzzles:", len(df))
print("Columns:", list(df.columns))
df.head()
```

```python
seed=42
test=1_000
validation=10_000
pool=100_000
total=pool+test+validation
sample=df.sample(n=total,random_state=seed).reset_index(drop=True)
print("sample size:",len(sample))
```

```python
def text_to_numbers(text):
    numbers=[]
    for ch in text:
        if ch =='.':
            numbers.append(0)
        else:
            numbers.append(int(ch))
    return numbers

print(text_to_numbers(sample["puzzle"][0]))
```

```python
import numpy as np
```

```python
puzzles = []
solutions = []

for i in range(len(sample)):
    puzzles.append(text_to_numbers(sample["puzzle"][i]))
    solutions.append(text_to_numbers(sample["solution"][i]))

puzzles = np.array(puzzles)
solutions = np.array(solutions)

print("puzzles:", puzzles.shape)
print("solutions:", solutions.shape)
```

```python
print(puzzles.shape)
print(puzzles[:100].shape)
print(puzzles[0].shape)
print(puzzles[0].reshape(9, 9).shape)
```

```python
def show_grid(cells):
    grid = cells.reshape(9, 9)
    for r in range(9):
        if r == 3 or r == 6:
            print("------+-------+------")
        row_text = ""
        for c in range(9):
            if c == 3 or c == 6:
                row_text += "| "
            if grid[r][c] == 0:
                row_text += ". "
            else:
                row_text += str(grid[r][c]) + " "
        print(row_text)

print("PUZZLE")
show_grid(puzzles[0])
print()
print("SOLUTION")
show_grid(solutions[0])
```

```python
def is_valid_solution(cells):
    grid = cells.reshape(9, 9)
    correct = {1, 2, 3, 4, 5, 6, 7, 8, 9}

    # check every row
    for r in range(9):
        if set(grid[r, :].tolist()) != correct:
            return False

    # check every column
    for c in range(9):
        if set(grid[:, c].tolist()) != correct:
            return False

    # check every 3x3 box
    for box_r in [0, 3, 6]:
        for box_c in [0, 3, 6]:
            box = grid[box_r:box_r+3, box_c:box_c+3]
            if set(box.flatten().tolist()) != correct:
                return False

    return True
```

```python
broken = solutions[0].copy()
broken[0] = broken[1]

print("Real solution:", is_valid_solution(solutions[0]))
print("Broken solution:", is_valid_solution(broken))
```

```python
count = 0
for s in solutions:
    if is_valid_solution(s):
        count += 1

print("Valid solutions:", count, "out of", len(solutions))
```

```python
def solution_matches_puzzle(puzzle, solution):
    for i in range(81):
        if puzzle[i] != 0 and puzzle[i] != solution[i]:
            return False
    return True

count = 0
for i in range(len(puzzles)):
    if solution_matches_puzzle(puzzles[i], solutions[i]):
        count += 1

print("Matching:", count, "out of", len(puzzles))
```

```python
rating = sample["difficulty"].astype(float)

print(rating.describe())
print()
print("Rating 0 (very easy):", (rating == 0).sum())
```

```python
groups = []
for r in rating:
    if r == 0:
        groups.append("easy")
    elif r <= 2:
        groups.append("medium")
    else:
        groups.append("hard")

groups = np.array(groups)

for g in ["easy", "medium", "hard"]:
    print(g, (groups == g).sum())
```

```python
rng = np.random.default_rng(seed)

test_idx = []
for g in ["easy", "medium", "hard"]:
    positions = np.where(groups == g)[0]
    chosen = rng.choice(positions, size=1000, replace=False)
    test_idx.extend(chosen)

test_idx = np.array(test_idx)
print("Test size:", len(test_idx))
```

```python
test_set = set(test_idx.tolist())

rest_idx = []
for i in range(len(puzzles)):
    if i not in test_set:
        rest_idx.append(i)

rest_idx = np.array(rest_idx)
val_idx = rest_idx[:10000]
pool_idx = rest_idx[10000:]

print("Test:", len(test_idx))
print("Validation:", len(val_idx))
print("Pool:", len(pool_idx))
print()
for g in ["easy", "medium", "hard"]:
    print("Test", g, (groups[test_idx] == g).sum())
```

```python
rating_array = rating.to_numpy()

def save_split(name, idx):
    np.savez_compressed(
        "/kaggle/working/" + name + ".npz",
        puzzles=puzzles[idx],
        solutions=solutions[idx],
        rating=rating_array[idx],
        groups=groups[idx],
    )
    print("Saved", name, "-", len(idx), "puzzles")

save_split("test", test_idx)
save_split("val", val_idx)
save_split("pool", pool_idx)
```

```python
puzzle_texts = sample["puzzle"].to_numpy()

test_texts = set(puzzle_texts[test_idx])
val_texts = set(puzzle_texts[val_idx])
pool_texts = set(puzzle_texts[pool_idx])

print("Test also in pool:", len(test_texts & pool_texts))
print("Test also in val:", len(test_texts & val_texts))
print("Val also in pool:", len(val_texts & pool_texts))
```
